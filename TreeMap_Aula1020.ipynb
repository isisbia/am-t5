# Pipeline | Construção do Algoritmo ML - TreeMap [Credit]

```python
!pip -q install yellowbrick
!pip -q install scikit-learn
!pip -q install matplotlib
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
arvore_credit = DecisionTreeClassifier(criterion='entropy', random_state=0)
arvore_credit.fit(X_credit_train, y_credit_train)
```

```python
DecisionTreeClassifier(ccp_alpha=0.0, class_weight=None, criterion='entropy',
                      max_depth=None, max_features=None, max_leaf_nodes=None,
                      min_impurity_decrease=0.0,
                      min_samples_leaf=1, min_samples_split=2,
                      min_weight_fraction_leaf=0.0,
                      random_state=0, splitter='best')
```

```python
tree_predicao = arvore_credit.predict(X_credit_test)
tree_predicao
```

```python
y_credit_test
```

```python
from sklearn.metrics import accuracy_score, classification_report
accuracy_score(y_credit_test, tree_predicao)
```

```python
from yellowbrick.classifier import ConfusionMatrix
cm = ConfusionMatrix(arvore_credit)
cm.fit(X_credit_train, y_credit_train)
cm.score(X_credit_test, y_credit_test)
```

```python
print(classification_report(y_credit_test, tree_predicao))
```

# Pipeline | RandomForest [Credit]

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
random_forest_credit = RandomForestClassifier(n_estimators=40, criterion='entropy', random_state=0)
random_forest_credit.fit(X_credit_train, y_credit_train)
```

```python
rf_predicoes = random_forest_credit.predict(X_credit_test)
rf_predicoes
```

```python
accuracy_score(y_credit_test, rf_predicoes)
```

```python
rf_cm = ConfusionMatrix(random_forest_credit)
rf_cm.fit(X_credit_train, y_credit_train)
rf_cm.score(X_credit_test, y_credit_test)
```

```python
print(classification_report(y_credit_test, rf_predicoes))
```
